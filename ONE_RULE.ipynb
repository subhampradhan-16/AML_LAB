```python
import pandas as pd
import numpy as np
```

```python
dataset={
    "Rain_Fall":["High","High","High","Low","Low","Low"],
    "Soil_Type":["Loamy","Sandy","Clay","Loamy","Sandy","Clay"],
    "Yield":["Good","Good","Good","Poor","Poor","Poor",]
}
data=pd.DataFrame(dataset)
```

```python
data
```

```python
def oner(data,target):
    feature_error={}
    feature_total_error={}
    temp=data.drop(target,axis=1)
    features=list(temp.columns)
    for feature in features:
        total_error=0
        print(f"feature = {feature}\nUnique Values={data[feature].unique()}\n\n")
        for val in data[feature].unique():
            val_target=data[data[feature]==val][target]
            majority_class=val_target.mode()[0]
            total_error+=sum(majority_class != val_target)
        e=total_error/data[feature].shape[0]
        feature_error[feature]=e
        feature_total_error[feature]=total_error
    return feature_error,feature_total_error
```

```python
feature_error,feature_total_error=oner(data,"Yield")

for feature,error in feature_error.items():
    print(f"Feature = {feature} \nError={error}\nMisclassified={feature_total_error[feature]}\n\n")
sorted_feature_error=dict(sorted(feature_error.items(),key=lambda item:item[1]))
```

```python
selected_feature=next(iter(sorted_feature_error))

print(f"The {selected_feature} feature is selected because it have less error rate of {feature_error[selected_feature]} .")
```
